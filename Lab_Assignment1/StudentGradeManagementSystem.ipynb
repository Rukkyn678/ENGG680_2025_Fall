```python
# step 5
# define a curve fitting function named fit_polynomial_curve
# the function should implement the linear least square estimation
# the function should include all important intermediate steps in LSE
# the input and output of the funciton are described in the comment below in red.
# please keep code clean and well-commented
# (7 marks)

def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients
        apv:            a posterier variance factor (goodness of fit)
        v:              observation residuals
    '''


    # Number of observations
    n = len(a)

    # Step 1: Construct the design matrix (Vandermonde matrix) for polynomial fitting
    # Each column is a^0, a^1, ..., a^deg
    A = np.vander(a, N=deg+1, increasing=True)  # shape: (n, deg+1)

    # Step 2: Calculate the matrix to invert (normal equation matrix)
    # M = A^T * P * A
    M = A.T @ P @ A

    # Step 3: Calculate the right-hand side vector
    # b = A^T * P * l
    b = A.T @ P @ l

    # Step 4: Solve for polynomial coefficients: x = M^{-1} * b
    x = np.linalg.solve(M, b)

    # Step 5: Calculate the covariance matrix of estimated coefficients: cov_x = sigma^2 * M^{-1}
    # For now, sigma^2 is unknown, so cov_x = M^{-1} scaled later
    cov_x = np.linalg.inv(M)

    # Step 6: Compute observation residuals: v = l - A * x
    v = l - A @ x

    # Step 7: Calculate a posteriori variance factor (apv)
    # apv = (v^T * P * v) / (n - (deg + 1))
    dof = n - (deg + 1)  # degrees of freedom
    apv = (v.T @ P @ v) / dof

    # Scale covariance matrix by apv to reflect uncertainty properly
    cov_x = cov_x * apv

    return (x, cov_x, apv, v)


```

```python
# step 6
# call the fit_polynomial_curve function and fetch the results
# use polynomial degrees of: 2, 4, 6
# (1 marks)
degrees = [2, 4, 6]

results = {}

for deg in degrees:
    x, cov_x, apv, v = fit_polynomial_curve(a, l, P, deg)
    results[deg] = {
        'coefficients': x,
        'covariance': cov_x,
        'apv': apv,
        'residuals': v
    }
    print(f"Degree {deg} polynomial fit done.")
```

```python
# step 7
# use the estimated coefficients x and constants array a, construct polynomial functions for each degree
# hint: use polyval function from numpy library (check input argument carefully)
# (1 marks)

x2 = results[2]['coefficients']
x4 = results[4]['coefficients']
x6 = results[6]['coefficients']

# np.polyval expects coefficients in descending order, so reverse them
quadratic_uw = np.polyval(x2[::-1], a)
quartic_uw = np.polyval(x4[::-1], a)
hexic_uw = np.polyval(x6[::-1], a)
```

```python
# step 8
# use the matplotlib to plot each polynomial function, noisy observations l, true observations l_true, together in one single figure
# for the polynomials of line, cubic and quintic, use color of orange, green and red respectively.
# (2 marks)

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(14, 5))


# Scatter plot of noisy observations
ax.scatter(a, l, label='Noisy observations', color='tab:blue', marker='o', alpha=0.3)

# Plot true observations (line)
ax.plot(a, l_true, label='True observations', color='k')

# Plot polynomial fits
ax.plot(a, quadratic_uw, label='Quadratic (deg 2)', color='orange')
ax.plot(a, quartic_uw, label='Quartic (deg 4)', color='green')
ax.plot(a, hexic_uw, label='Hexic (deg 6)', color='red')

# Axis labels and title
ax.set_xlabel('a')
ax.set_ylabel('Observation Value')
ax.set_title('Polynomial Fits vs Observations')

# Legend
ax.legend()

# Limit x-axis
ax.set_xlim(-1, 1)

# Grid
ax.grid(True, which='both', ls='-', color='0.65')

fig.tight_layout()
plt.show()

```

```python
# step 9
# print the a posterier variance factor for each polynomial function
# (3 marks)
for deg in degrees:
    apv = results[deg]['apv']
    print(f"A posteriori variance factor for degree {deg}: {apv:.6f}")
```

```python
# step 10
# compute and print the norm between the predicted observations (output from each polynomial function) and the true observations
# (3 marks)

norm_quad = np.linalg.norm(quadratic_uw - l_true)
norm_quart = np.linalg.norm(quartic_uw - l_true)
norm_hexic = np.linalg.norm(hexic_uw - l_true)

# Print results
print(f"Norm between quadratic fit and true observations: {norm_quad:.6f}")
print(f"Norm between quartic fit and true observations: {norm_quart:.6f}")
print(f"Norm between hexic fit and true observations: {norm_hexic:.6f}")
```

```python
# step 12
# now, let's define the P matrix with the proper weights of observation
# use the variance of observation l_var, define P matrix
# (1 mark)
P = np.diag(1 / l_var)
```

```python
# step 13
# repeat the steps 6-10, using the proper P matrix
# (7 marks)
# Step 6: Call fit_polynomial_curve for degrees 2, 4, 6 using weighted P
degrees = [2, 4, 6]

results_weighted = {}

for deg in degrees:
    x, cov_x, apv, v = fit_polynomial_curve(a, l, P, deg)
    results_weighted[deg] = {
        'coefficients': x,
        'covariance': cov_x,
        'apv': apv,
        'residuals': v
    }
    print(f"Degree {deg} weighted polynomial fit done.")

# Step 7: Construct polynomial functions using weighted coefficients
quadratic_w = np.polyval(results_weighted[2]['coefficients'][::-1], a)
quartic_w = np.polyval(results_weighted[4]['coefficients'][::-1], a)
hexic_w = np.polyval(results_weighted[6]['coefficients'][::-1], a)

# Step 8: Plot the fits with noisy and true observations
fig, ax = plt.subplots(figsize=(14, 5))

ax.scatter(a, l, label='Noisy observations', color='tab:blue', marker='o', alpha=0.3)
ax.plot(a, l_true, label='True observations', color='k')

ax.plot(a, quadratic_w, label='Quadratic weighted (deg 2)', color='orange')
ax.plot(a, quartic_w, label='Quartic weighted (deg 4)', color='green')
ax.plot(a, hexic_w, label='Hexic weighted (deg 6)', color='red')

ax.set_xlabel('a')
ax.set_ylabel('Observation Value')
ax.set_title('Weighted Polynomial Fits vs Observations')
ax.legend()
ax.set_xlim(-1, 1)
ax.grid(True, which='both', ls='-', color='0.65')

fig.tight_layout()
plt.show()

# Step 9: Print the a posteriori variance factors for weighted fits
for deg in degrees:
    print(f"Weighted apv for degree {deg}: {results_weighted[deg]['apv']:.6f}")

# Step 10: Compute and print norms between weighted polynomial predictions and true observations
norm_quad_w = np.linalg.norm(quadratic_w - l_true)
norm_quart_w = np.linalg.norm(quartic_w - l_true)
norm_hexic_w = np.linalg.norm(hexic_w - l_true)

print(f"Norm between weighted quadratic fit and true observations: {norm_quad_w:.6f}")
print(f"Norm between weighted quartic fit and true observations: {norm_quart_w:.6f}")
print(f"Norm between weighted hexic fit and true observations: {norm_hexic_w:.6f}")

```
